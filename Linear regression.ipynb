```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

# Transaction amounts
amounts = np.array([
    100, 120, 110, 95, 130,
    105, 115, 125, 108, 98,
    5000, 4500
])

# Convert to 2D
X = amounts.reshape(-1, 1)

# Create model
model = IsolationForest(
    n_estimators=100,
    contamination=0.15,
    random_state=42
)

# Train
model.fit(X)

# Predict
predictions = model.predict(X)

# Display results
for amount, prediction in zip(amounts, predictions):
    if prediction == -1:
        print(amount, "→ Anomaly")
    else:
        print(amount, "→ Normal")

# Plot
plt.scatter(range(len(amounts)), amounts, c=predictions)
plt.xlabel("Transaction")
plt.ylabel("Amount")
plt.title("Isolation Forest Anomaly Detection")
plt.show()
```
